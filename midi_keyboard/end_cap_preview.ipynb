```python
from jupyterscad import view

from end_cap import EndCapBuilder
from octave import OctaveBuilder
from configuration import load_config
from backplate import BackplateBuilder
```

```python
conf = load_config()
conf
```

```python
octave = OctaveBuilder(7, conf)
```

```python
end_cap = EndCapBuilder(octave, conf)
model = end_cap.build()
backplate = BackplateBuilder(end_cap, conf)
backplate.move([0, 0, -5])
view(model + backplate.build())
```

```python
conf.output_dir.mkdir(parents=True, exist_ok=True)
model.save_as_stl(conf.output_dir / "end_cap_v1.stl")
```
