```python
from jupyterscad import view
from solid2 import square, circle
from solid2.extensions.bosl2 import round_corners
from solid2 import polygon
```

```python
U_WIDTH = 1.25
U_LEN = 1.75
```

```python
nozzle_size = 0.4
```

```python
mx_stem_l = 4.0
mx_stem_w = 1.2
mx_stem_h = 3.0
mx_stem_l, mx_stem_w, mx_stem_h
```

```python
mx_stem_arm = square([mx_stem_l, mx_stem_w], center=True)
mx_stem_base = circle(d=5.5) - (mx_stem_arm + mx_stem_arm.rotateZ(90))
```

```python
mx_stem = mx_stem_base.linear_extrude(height=mx_stem_h)
view(mx_stem)
```

```python
u = 18
keycap_h = 1
round_r = 2
wall_width = 1
wall_offset = wall_width * 2

keycap_base = square([u * U_LEN, u * U_WIDTH], center=True)
walls = keycap_base - square(
    [u * U_LEN - wall_offset, u * U_WIDTH - wall_offset], center=True
)
rounded_keycap = polygon(
    round_corners(path=keycap_base, radius=round_r)
).linear_extrude(height=keycap_h)
view(rounded_keycap)
```

```python
final = rounded_keycap + mx_stem.up(keycap_h - 0.01)
view(final)
```

```python
final.save_as_stl("sg_keycap.stl")
```
