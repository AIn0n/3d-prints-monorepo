```python
from jupyterscad import view
from solid2 import square
```

```python
u = 14
white_key_dist = 19.5 * 1.25
white_key_len = 19.5 * 1.75
plate_width = 1.5
```

```python
mx_hole = square([u, u], center=True)
mounting_plate = square([white_key_len, white_key_dist], center=True) - mx_hole
view(mounting_plate.linear_extrude(plate_width))
```
