```python
from jupyterscad import view
from solid2 import square
from solid2.extensions.bosl2 import hexagon
```

```python
HEX_R = 15
HEX_H = 8
MNT_SIZE = 5
```

```python
base = hexagon(ir=HEX_R).linear_extrude(HEX_H) 
mounting_square = square([MNT_SIZE,MNT_SIZE], center=True).linear_extrude(HEX_H - 1)
base = base + mounting_square.forward(HEX_R + MNT_SIZE // 2) - mounting_square.back(HEX_R - MNT_SIZE // 2)
base = base.rotateZ(120) + mounting_square.forward(HEX_R + MNT_SIZE // 2) - mounting_square.back(HEX_R - MNT_SIZE // 2)
base = base.rotateZ(120) + mounting_square.forward(HEX_R + MNT_SIZE // 2) - mounting_square.back(HEX_R - MNT_SIZE // 2)
view(base)
```

```python
base.save_as_stl("hex.stl")
```
