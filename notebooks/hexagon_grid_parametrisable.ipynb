```python
from jupyterscad import view
from solid2.extensions.bosl2 import hexagon, trapezoid
```

```python
#   base hexagon shape related:
#   where:
#       R   - base inner radius (half of the distance between two lines)
#       H   - total height of structure
HEX_R = 15
HEX_H = 8

#   Trapezoid mounting point related:
#   where:
#       SIZE    - length of the mounting point, over the countour of hex
#               - width of the trapezoid's smaller base
#       H       - total height of mounting point
#       BASE_DIFF   - difference between smaller and bigger trapezoid's base
#                   - It's added to the SIZE constant
#       H_MARGIN    - margin of the height - the amount of height cut off
#                     from _male_ mounting point. This way it's way easier
#                     to get flat surface on the top
#       SIDE_MARGIN - margin from the sides of _male_mounting point.
MNT_SIZE = 5
MNT_H = HEX_H - 1

MNT_BASE_DIFF = 2
MNT_H_MARGIN = 1
MNT_SIDE_MARGIN = 0.15
```

```python
mounting_male_len = MNT_SIZE - MNT_SIDE_MARGIN
mounting = (
    trapezoid(
        h=mounting_male_len,
        w1=MNT_SIZE - MNT_SIDE_MARGIN,
        w2=MNT_SIZE + MNT_BASE_DIFF - MNT_SIDE_MARGIN,
        center=True,
    )
    .linear_extrude(MNT_H - MNT_H_MARGIN)
    .forward(HEX_R + mounting_male_len // 2)
)
hole = (
    trapezoid(h=MNT_SIZE, w1=MNT_SIZE, w2=MNT_SIZE + MNT_BASE_DIFF, center=True)
    .linear_extrude(MNT_H)
    .back(HEX_R - MNT_SIZE // 2)
)


def add_mounting_holes(base_shape):
    return base_shape + mounting - hole
```

```python
base = hexagon(ir=HEX_R).linear_extrude(HEX_H)

base = add_mounting_holes(base)
base = add_mounting_holes(base.rotateZ(60))
base = add_mounting_holes(base.rotateZ(60))
view(base)
```

```python
base.save_as_stl("hex.stl")
```
